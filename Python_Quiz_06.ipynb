```python
# Task 1: E-Commerce Customer Analytics & Discount Engine


# 1. Create a list of dictionaries containing customer transactions.

transactions = [
    {"Customer_Name": "Ahmed", "Category": "Electronics", "Items_Bought": 2, "Total_Spent": 650.0},
    {"Customer_Name": "Sara", "Category": "Clothing", "Items_Bought": 5, "Total_Spent": 320.0},
    {"Customer_Name": "Omar", "Category": "Electronics", "Items_Bought": 1, "Total_Spent": 1200.0},
    {"Customer_Name": "Mona", "Category": "Home", "Items_Bought": 3, "Total_Spent": 450.0},
    {"Customer_Name": "Khaled", "Category": "Clothing", "Items_Bought": 4, "Total_Spent": 550.0}
]


# 2. Create a function named analyze_customer_spending(transactions).
# Calculate total revenue and find the customer who spent the highest amount.

def analyze_customer_spending(transactions):

    total_revenue = 0
    highest_spending = 0
    top_spender = ""

    for customer in transactions:

        total_revenue += customer["Total_Spent"]

        if customer["Total_Spent"] > highest_spending:
            highest_spending = customer["Total_Spent"]
            top_spender = customer["Customer_Name"]

    return total_revenue, top_spender, highest_spending


# Call the function and print the results.

total_revenue, top_spender, highest_spending = analyze_customer_spending(transactions)

print("Total Revenue: $", total_revenue)
print("Top Spender:", top_spender, "($", highest_spending, ")")


# 3. Use Dictionary Comprehension to create a new dictionary.
# Keep customers whose spending is greater than $500.
# Apply a 10% loyalty discount to their spending amounts.

vip_customers = {
    customer["Customer_Name"]: customer["Total_Spent"] * 0.90
    for customer in transactions
    if customer["Total_Spent"] > 500
}


# Print VIP customers with discounted totals.

print("\nVIP Customers with Discounted Totals (Spent > $500):")
print(vip_customers)
```

```python
# Task 2: Employee Performance & Salary Simulation (NumPy & Pandas)

import numpy as np
import pandas as pd


# 1. NumPy Vectorization:
# Create a NumPy array containing monthly base salaries for 6 employees.

base_salaries = np.array([5000, 7500, 4200, 9000, 6100, 8000])


# Calculate a New Salary with a 12% performance bonus.
# Round the results to 2 decimal places.

new_salaries = np.round(base_salaries * 1.12, 2)


# 2. Pandas DataFrame Creation:
# Create a dictionary containing employee details.

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Ali", "Fatma", "Hassan", "Zainab", "Ibrahim", "Nour"],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR"],
    "Experience_Years": [2, 5, 1, 6, 3, 4]
}


# Create a Pandas DataFrame.

df = pd.DataFrame(data)


# Add the calculated salaries to the DataFrame.

df["Updated_Salary"] = new_salaries


# 3. Filtering & Sorting:
# Filter employees with 3 or more years of experience.

filtered_df = df[df["Experience_Years"] >= 3]


# Sort employees by Updated_Salary in descending order.

sorted_df = filtered_df.sort_values(
    by="Updated_Salary",
    ascending=False
)


# Print the final results.

print("=== Filtered & Sorted Experienced Employees (Experience >= 3) ===")
print(sorted_df)
```
